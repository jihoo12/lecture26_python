```python
#while 문 연습: 나무 찍기
tree_hit = 1
while tree_hit < 10:
    print(f'나무를 {tree_hit}번 찍었습니다.')
    tree_hit += 1
print("나무가 넘어갑니다.")
```

```python
#while문 연습:입력값에 따라 반복 종료하기
sum = count = 0
num = int(input("숫자 입력 (0: 끝) :"))
while num != 0:
    sum += num
    count += 1
    num = int(input("숫자 입력 (0: 끝):"))
if count != 0:
    average = sum / count
else:
    print(0)
print(f">> 총합 : {sum},평균: {average:.2f}")
```

```python
#무한루프와 분기문 break활용
sum = count = 0
while True:
    num = int(input("숫자 입력 (0: 끝):"))
    if num== 0:
        break
    sum += num
    count += 1
if count != 0:
    average = sum / count
else:
    print(0)
print(f">> 총합 : {sum},평균: {average}")
```
