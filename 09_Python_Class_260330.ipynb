### python class

```python
class Circle:
    pi = 3.14
    def __init__(self,radius):
        self.radius = radius
    def area():
        return self.radius ** 2 * Circle.pi
c = Circle(2.5)
```

```python
class Rectangle:
    def __init__(self,width,height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
rec = Rectangle(2.1,3.4)
```

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def __str__(self):
        return f'{self.name}({self.age}살)'
    def __eq__(self,other):
        if self.name == other.name and self.age == other.age:
            return True
        return False
p1 = Person('홍길동',15)
print(p1)
p2 = Person('홍길동',20)
if p1 == p2:
    print('같은 사람')
else:
    print('다른 사람')
```
