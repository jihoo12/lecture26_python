### python class

```python
class Circle:
    pi = 3.14
    def __init__(self,radius):
        self.radius = radius
    def area():
        return self.radius ** 2 * Circle.pi
c = Circle(2.5)
```

```python
class Rectangle:
    def __init__(self,width,height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
rec = Rectangle(2.1,3.4)
```
