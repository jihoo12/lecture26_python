```python
try:
    num_list = [2, 6, 34, 65, 11]
    index = int(input('인덱스 입력:'))
    print(num_list[index])
except ValueError:
    print(f'0부터 {len(num_list)-1}까지의 정수를 입력하세요.')
except IndexError:
    print(f'0부터 {len(num_list)-1}까지의 정수를 입력하세요.')
except Exception as e:
    print(f'예외가 발생했습니다 : {e} -> {type(e)}')
```

```python
while True:
    try:
        number = int(input("양의 정수 입력 : "))
        if number <= 0:
            raise ValueError('양의 정수를 입력하세요')
    except ValueError:
            print("양의 정수를 입력하세요.")
    else:
        sum = 0
        for i in range(1,number + 1):
            sum += i
            print(f'>> 1부터 {number}까지의 합 : {sum}')
    break
```

```python
def divide(x, y):
    try:
        # 1. 시도할 연산
        result = x / y
    except ZeroDivisionError:
        # 2. 0으로 나누기 예외가 발생했을 때 처리
        print("0으로는 나눌 수 없습니다.")
    else:
        # 3. 예외가 발생하지 않았을 때 결과 출력
        print(f"결과: {result}")

# 다음 두 호출에 대한 결과 확인
divide(3.2, 2)
divide(5.4, 0)
```

```python
uyear = {1: 'freshman', 2: 'sophomore', 3: 'junior', 4: 'senior'}

try:
    # 1. 사용자로부터 정수 입력 받기
    choice = int(input("대학교 몇 학년이지요? "))
    
    # 2. 1~4 사이를 벗어나면 의도적으로 예외(Exception) 발생
    if choice not in uyear:
        raise Exception("1~4 정수를 입력하세요.")
        
except Exception as e:
    # 3. 예외가 발생했을 때 실행되는 구간
    print(f"예외 발생 이름: {type(e)}")
    print(f"예외 발생 이유: {e}")
    
else:
    # 4. 예외가 발생하지 않았을 때 실행되는 구간
    print(f"{choice}학년: {uyear[choice]}")
    
finally:
    # 5. 예외 발생 여부와 상관없이 무조건 실행되는 구간
    print("예외 처리가 잘되는군요!")
```
