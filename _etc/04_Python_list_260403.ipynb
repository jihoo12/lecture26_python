```python
def getIndex(arr,target):
    data =0
    for i in arr:
        data = data+1
        if i == target:
            break
    data = data-1
    return data 
def getMax(arr):
    return max(arr)
def getMin(arr):
    return min(arr)
def countGT(arr,target):
    count = 0 
    for i in arr:
        if i> target:
            count = count+1
    return count
def sumList(arr):
    return sum(arr)
def swapList(arr):
    arr.revert()
    return arr
scores = [95,45,21,74,55]
print(getIndex(scores,55))
```
