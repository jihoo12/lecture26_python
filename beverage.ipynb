영수증 출력

```python
import datetime

# 1. 메뉴 설정 (음료 이름: 가격)
menu = {
    "아메리카노": 3000,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "아이스티": 3000,
    "자몽에이드": 4500
}

def run_kiosk():
    # 메뉴 출력
    print("="*30)
    print("      CAFE PYTHON MENU      ")
    print("="*30)
    for name, price in menu.items():
        print(f"{name:<10} | {price:>5}원")
    print("="*30)

    # 2. 주문 받기
    while True:
        choice = input("\n주문하실 음료 이름을 입력하세요: ").strip()
        if choice in menu:
            price = menu[choice]
            break
        else:
            print("❌ 메뉴에 없는 음료입니다. 다시 입력해주세요.")

    # 3. 금액 출력 및 돈 받기
    print(f"\n> {choice}를 선택하셨습니다. 결제 금액은 {price}원입니다.")
    
    while True:
        try:
            money_in = int(input("지불할 금액을 입력하세요: "))
            if money_in >= price:
                break
            else:
                print(f"❌ 금액이 부족합니다. (부족한 금액: {price - money_in}원)")
        except ValueError:
            print("❌ 숫자만 입력 가능합니다.")

    # 4. 거스름돈 계산
    change = money_in - price
    print(f"\n✅ 결제가 완료되었습니다. 거스름돈은 {change}원입니다.")

    # 5. 영수증 출력
    print("\n" + "*"*30)
    print("         RECEIPT         ")
    print("*"*30)
    print(f"일시: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print(f"상품명: {choice}")
    print(f"결제금액: {price:>15}원")
    print(f"받은금액: {money_in:>15}원")
    print("-" * 30)
    print(f"거스름돈: {change:>15}원")
    print("*"*30)
    print("      감사합니다! 또 오세요!     ")

# 프로그램 실행
if __name__ == "__main__":
    run_kiosk()
```
