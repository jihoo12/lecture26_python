```python
import math 
import random
def getarea(r):
    return round(math.pi,4) * r**2
r = random.random()*10
print("원의 반지름:",r)
print("원주율 pi:", round(math.pi,4))
print(f"반지름 {r}인 원의 면적은 {getarea(r)}")
```

```python
import math
import statistics as s
def print_factorial():
    print(f"1! = {math.factorial(1)}")
    print(f"6! = {math.factorial(6)}")
    print(f"11! = {math.factorial(11)}")
    print(f"16! = {math.factorial(16)}")
def print_mmv():
    st = [80,99,77,65,92,74,82]
    print(st)
    print(f"중앙 값: {s.median(st)}")
    print(f"평균: {s.mean(st)}")
    print(f"중앙 값: {s.variance(st)}")
    print(f"중앙 값: {s.stdev(st)}")

if __name__ == '__main__':
    print_factorial()
    print_mmv()
```
